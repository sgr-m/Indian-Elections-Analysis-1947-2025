### WORKING WITH ELECTION RESULT 1951 - 2019 (CSV FILE)

#### 1. CLEANING THE YEAR

```python
# The 'election_year' looks like "1951 (1st LOK SABHA)" -> Unformatted
# Using a regular expression (regex) to extract only the 4 digits.
# Alternatively we can using string methods to do that and typecast to int but it will be very messy
def clean_election_year(year_val):
    if pd.isna(year_val):
        return None
    # Extracting the first 4-digit sequence found in the string
    match = re.search(r'\d{4}', str(year_val))
    return int(match.group()) if match else None

# Applying the cleaning function to create a new numeric year column.
election_results_1951_2019['year_int'] = election_results_1951_2019['election_year'].apply(clean_election_year)
election_results_1951_2019.head()
```

```python
election_results_1951_2019['state'].unique()
```

#### 2. ISOLATING UNIQUE HISTORICAL STATES

```python
# This captures every variation like 'Dilli', 'Mysore', or 'Madras'[cite: 16].
unique_historical_states = sorted(election_results_1951_2019['state'].unique().tolist())

print(f"Total Unique State Entities: {len(unique_historical_states)}")
print("--- Full State List for Mapping ---")
print(unique_historical_states)
```

#### 3. CREATING THE CROSS-TABULATION (Timeline Analysis)

```python
# This shows exactly which years 'Hyderabad' or 'Bombay' appear in the data[cite: 176].
# It helps you verify the 'cutoff' years for state bifurcations.
state_timeline = pd.crosstab(election_results_1951_2019['state'], election_results_1951_2019['year_int'])
state_timeline
```

#### Removing Stupid Hurdles 

```python
# 1. Normalize Case first to reduce the number of unique values
election_results_1951_2019['state_upper'] = election_results_1951_2019['state'].str.upper().str.strip()
```

```python
# 2. Comprehensive Mapping Dictionary
# This maps historical names, spelling variations, and UT changes to modern names.
state_standardization_map = {
    # Spelling & Case Variations
    'ORISSA': 'ODISHA',
    'PONDICHERRY': 'PUDUCHERRY',
    'UTTARANCHAL': 'UTTARAKHAND',
    'UTTARKHAND': 'UTTARAKHAND',
    'CHATTISGARH': 'CHHATTISGARH',
    'NATIONAL CAPITAL TERRITORY OF DELHI': 'DELHI',
    'LACCADIVE, MINICOY AND AMINDIVI ISLANDS': 'LAKSHADWEEP',
    
    # Historical Reorganizations (1951-1956 Era)
    'HYDERABAD': 'TELANGANA', # Historical core became part of modern Telangana
    'BOMBAY': 'MAHARASHTRA',  # Note: Bombay split into MH and GJ in 1960
    'MYSORE': 'KARNATAKA',
    'MADRAS': 'TAMIL NADU',
    'TRAVANCORE COCHIN': 'KERALA',
    'SAURASHTRA': 'GUJARAT',
    'KUTCH': 'GUJARAT',
    'MADHYA BHARAT': 'MADHYA PRADESH',
    'VINDHYA PRADESH': 'MADHYA PRADESH',
    'BHOPAL': 'MADHYA PRADESH',
    'AJMER': 'RAJASTHAN',
    'COORG': 'KARNATAKA',
    'PATIALA AND EAST PUNJ': 'PUNJAB',
    'GOA DAMAN AND DIU': 'GOA' # Separated later, but helps group early data
}
```

```python
# 3. Create the 'standardized_state' column
# We use the map; if a state isn't in the map, we keep the original upper-case name.
election_results_1951_2019['standardized_state'] = election_results_1951_2019['state_upper'].replace(state_standardization_map)
```

```python
election_results_1951_2019.head()
```

```python
# 4. Handle the Parent-Child Mapping for 2000/2014 Bifurcations
# To calculate long-term volatility, we need a 'parent_state' column.
parent_state_map = {
    'UTTARAKHAND': 'UTTAR PRADESH',
    'JHARKHAND': 'BIHAR',
    'CHHATTISGARH': 'MADHYA PRADESH',
    'TELANGANA': 'ANDHRA PRADESH'
}

election_results_1951_2019['parent_state'] = election_results_1951_2019['standardized_state'].replace(parent_state_map)

print("Standardization Complete.")
print(f"Original Unique States: {len(unique_historical_states)}")
print(f"Standardized Unique States: {election_results_1951_2019['standardized_state'].nunique()}")
```

#### 1. FIXING THE BOMBAY/GUJARAT GEOGRAPHIC TRAP

```python

# We target specific districts/constituencies that were in Bombay State 
# in 1951 but are now in Gujarat to ensure Chapter 5 heatmaps are correct.
gujarat_constituencies = ['AHMEDABAD', 'MEHSANA', 'AMRELI', 'PANCH MAHALS', 'KAIRA']

def fix_bombay_split(row):
    # If the state is marked as Maharashtra (from our previous map) 
    # but the constituency is historically in Gujarat, we fix it.
    if row['standardized_state'] == 'MAHARASHTRA' and any(cn in row['constitution'].upper() for cn in gujarat_constituencies):
        return 'GUJARAT'
    return row['standardized_state']

election_results_1951_2019['standardized_state'] = election_results_1951_2019.apply(fix_bombay_split, axis=1)
election_results_1951_2019.head(20)
```

#### 2. CLEANING NUMERICAL STRINGS

```python
# Removing '%' and commas to allow for Chapter 6 index calculations.
election_results_1951_2019['votes_received_perc'] = (
    election_results_1951_2019['votes_received_perc']
    .astype(str)
    .str.replace('%', '', regex=False)
    .astype(float)
)

# Converting votes to numeric, handling potential commas in the string data.
election_results_1951_2019['votes_received'] = pd.to_numeric(
    election_results_1951_2019['votes_received'], errors='coerce'
).fillna(0).astype(int)

election_results_1951_2019.votes_received.head()
```

#### 3. CREATING A BINARY WINNER FLAG

```python
# Essential for Random Forest win/loss prediction.
election_results_1951_2019['is_winner'] = (
    election_results_1951_2019['result'].str.strip().str.lower() == 'winner'
).astype(int)
```

```python
# 4. MULTI-MEMBER CONSTITUENCY IDENTIFIER
# Helping Chapter 3 SQL queries identify 1951/1957 multi-seat logic.
election_results_1951_2019['is_multimember'] = (election_results_1951_2019['num_seats'] > 1).astype(int)
```

## Minor fixes

```python
# 1. FIXING THE ID COLUMN (Primary Key Preparation)
# the original 'id' has 11 nulls, we'll replace it with a clean, unique integer range to act as your MySQL Primary Key.
election_results_1951_2019['id'] = range(1, len(election_results_1951_2019) + 1)
```

```python
# 2. FILLING NUMERICAL NaNs
# Filling missing vote percentages with 0 to ensure the ENP and 
# Volatility formulas don't return 'NaN'.
election_results_1951_2019['votes_received_perc'] = election_results_1951_2019['votes_received_perc'].fillna(0.0)
```

```python
# 3. HANDLING MISSING CATEGORICAL DATA 
# Using placeholders for gender, age, and category to prevent SQL 'NOT NULL' errors.
election_results_1951_2019['gender'] = election_results_1951_2019['gender'].fillna('UNKNOWN')
election_results_1951_2019['age'] = election_results_1951_2019['age'].fillna(0)

# I am taking General (may affect little bit significance --> will see it later)
election_results_1951_2019['category'] = election_results_1951_2019['category'].fillna('GENERAL')
```

```python
# 4. REFINING PARENT-STATE FOR THE BOMBAY ERA
# For the 1951 and 1957 elections, both Gujarat and Maharashtra constituencies 
# should point to 'BOMBAY' as the parent for accurate longitudinal modeling.
def refine_parent_logic(row):
    if row['year_int'] <= 1960 and row['standardized_state'] in ['GUJARAT', 'MAHARASHTRA']:
        return 'BOMBAY'
    return row['parent_state']

election_results_1951_2019['parent_state'] = election_results_1951_2019.apply(refine_parent_logic, axis=1)

print("Final Dataframe fixes applied successfully.")
election_results_1951_2019.info()
```

### Creating SQL file...

```python
# Exporting the 1951-2019 master data
election_results_1951_2019.to_sql(
    name='election_master',      # Name of the table in MySQL
    con=engine,                  # The engine we created above
    if_exists='replace',         # 'replace' creates the table; 'append' adds to it
    index=False,                 # Do not export the Pandas index as a column
    chunksize=1000               # Sends 5000 rows at a time for efficiency
)

print("Export to MySQL Complete!")
```
